### Registering dataset

```python
from azure.ai.ml import MLClient
from azure.ai.ml.entities import Data
from azure.ai.ml.constants import AssetTypes
from azure.identity import InteractiveBrowserCredential

# Instantiate MLClient
credential = InteractiveBrowserCredential()
ml_client = MLClient(
    credential=credential,
    subscription_id="0a94de80-6d3b-49f2-b3e9-ec5818862801",
    resource_group_name="buas-y2",
    workspace_name="NLP9-2025"
)

def register_data_asset(name, path, version="1"):
    """
    Registers a data asset in Azure ML workspace.

    Args:
        name (str): Name of the data asset.
        path (str): Path to the data file (local or remote).
        version (str, optional): Version of the data asset. Defaults to "1".

    Returns:
        None
    """
    data_asset = Data(
        path=path,  # Local or remote path to the file
        type=AssetTypes.URI_FILE,
        description=f"{name} (v{version})",
        name=name,
        version=version
    )
    ml_client.data.create_or_update(data_asset)

register_data_asset("processed_dataset", "azureml://datastores/workspaceblobstore/paths/processed_dataset/processed_dataset.csv")
register_data_asset("subset_1", "azureml://datastores/workspaceblobstore/paths/processed_dataset/subset_1.csv")
register_data_asset("subset_2", "azureml://datastores/workspaceblobstore/paths/processed_dataset/subset_2.csv")
register_data_asset("subset_3", "azureml://datastores/workspaceblobstore/paths/processed_dataset/subset_3.csv")
register_data_asset("subset_4", "azureml://datastores/workspaceblobstore/paths/processed_dataset/subset_4.csv")
register_data_asset("subset_5", "azureml://datastores/workspaceblobstore/paths/processed_dataset/subset_5.csv")
```
